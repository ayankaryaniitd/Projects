```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
```

```python
def Deutsch(func):
    circuit= QuantumCircuit(2,1)
    circuit.x(1)
    circuit.h(0)
    circuit.h(1)
    circuit.compose(func, inplace=True)
    circuit.h(0)
    circuit.measure(0,0)
    sim= AerSimulator()
    job= sim.run(circuit, shots= 1024)
    results= job.result()
    print('Circuit Diagram')
    print(circuit)
    print('Result of the experiment:', results.get_counts())
```

```python
def Deutsch_Jozsa(func, n):
    circuit= QuantumCircuit(n+1, n)
    circuit.x(-1)
    for i in range(n+1):
        circuit.h(i)
    circuit.compose(func, inplace=True)
    for i in range(n):
        circuit.h(i)
    for i in range(n):
        circuit.measure(i,i)
    str= 0*n
    sim= AerSimulator()
    job= sim.run(circuit, shots= 1024)
    results= job.result()
    print('QuantumCircuit:')
    print(circuit)
    print('Result of the experiment:', results.get_counts())
    
```
